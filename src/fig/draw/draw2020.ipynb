```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import latexify
```

```python
@latexify.function
def f(x):
    return 1/(1+np.cos(x))**2
```

```python
x = np.linspace(0,np.pi/2, 1000)
dx = x[1] - x[0]
S = 0
for i in x:
    S += f(i)*dx

print(S)
```

```python
def f(x, a, b):
    return b/a * np.sqrt(a**2 - x**2)

a = 5
b = 3
x = np.linspace(-a, a, 1000)
yb = -f(x, a, b)
yu =  f(x, a, b)

fig,ax = plt.subplots()

ax.plot(x,yu,color='k',lw=0.8) # ,label=''
ax.plot(x,yb,color='k',lw=0.8) # ,label=''


# ax.set_xlabel('')

# ax.grid()
# ax.legend()
# ax.set_yscale('log')
ax.set_aspect('equal')
ax.plot([0,0],[-(b+1),(b+1)],lw=0.5,color='k')
ax.plot([-(a+1),(a+1)],[0,0],lw=0.5,color='k')


# plt.show()の前に入れないと白紙になる
plt.savefig('B_2020_ellipse.svg', dpi=1200, bbox_inches='tight', pad_inches=0.1)

plt.show()
```
